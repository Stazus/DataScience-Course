```python

from pathlib import Path

# Ścieżka do folderu na Dysku Google
folder = Path("/content/drive/MyDrive/DS_Course/Lesson02")

# Utworzenie folderu wraz z katalogami nadrzędnymi
folder.mkdir(parents=True, exist_ok=True)

# Sprawdzenie, czy folder istnieje
print("LEKCJA 2 – ZADANIE 2")
print("Tworzenie folderu na Dysku Google")
print("-" * 50)
print(f"Ścieżka: {folder}")
print(f"Czy folder istnieje? {folder.exists()}")
print(f"Czy jest katalogiem? {folder.is_dir()}")
```

```python

from datetime import date

# Plik tekstowy na Dysku Google
plik = folder / "informacje.txt"

# Zapisanie imienia i aktualnej daty
with open(plik, "w", encoding="utf-8") as f:
    f.write("Imię: Stanisław\n")
    f.write(f"Data: {date.today().isoformat()}\n")

print("Plik został zapisany.")
print(f"Ścieżka: {plik}")

print("\n=== ZAWARTOŚĆ FOLDERU ===")
for element in sorted(folder.iterdir()):
    print(f"- {element.name}")

print("\n=== ZAWARTOŚĆ PLIKU ===")
print(plik.read_text(encoding="utf-8"))
```
