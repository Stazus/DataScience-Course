```python
# ==================================================
# SEKCJA 3: TWORZENIE PRZYKŁADOWEGO DATASETU
# ==================================================

# Tworzymy zbiór danych sprzedażowych: 100 wierszy, 3 kolumny
df = pd.DataFrame({
    "produkt": np.random.choice(["Laptop", "Monitor", "Klawiatura"], size=100),
    "ilosc": np.random.randint(1, 11, size=100),
    "cena": np.random.randint(100, 5001, size=100)
})

print("Utworzono przykładowy dataset sprzedażowy.")
print(f"Liczba wierszy: {df.shape[0]}")
print(f"Liczba kolumn: {df.shape[1]}")

df.head()
```
