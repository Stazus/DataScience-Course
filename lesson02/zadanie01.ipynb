```python

import platform
import psutil
import sys

print("=" * 55)
print("LEKCJA 2 – ZADANIE 1")
print("INFORMACJE O ŚRODOWISKU DATA SCIENCE")
print("=" * 55)

print("\n=== SYSTEM OPERACYJNY ===")
print(f"System: {platform.system()}")
print(f"Wersja systemu: {platform.release()}")
print(f"Architektura: {platform.machine()}")
print(f"Python: {sys.version.split()[0]}")

print("\n=== PAMIĘĆ RAM ===")
ram = psutil.virtual_memory()
print(f"Pamięć całkowita: {ram.total / 1024**3:.2f} GB")
print(f"Pamięć dostępna: {ram.available / 1024**3:.2f} GB")
print(f"Wykorzystanie RAM: {ram.percent:.1f}%")

print("\n=== PROCESOR ===")
print(f"Rdzenie fizyczne: {psutil.cpu_count(logical=False)}")
print(f"Procesory logiczne: {psutil.cpu_count(logical=True)}")

print("\n=== KARTA GRAFICZNA GPU ===")
try:
    import torch

    if torch.cuda.is_available():
        print("CUDA dostępna: TAK")
        print(f"GPU: {torch.cuda.get_device_name(0)}")
        gpu = torch.cuda.get_device_properties(0)
        print(f"Pamięć GPU: {gpu.total_memory / 1024**3:.2f} GB")
    else:
        print("CUDA dostępna: NIE")
except ImportError:
    print("Nie można sprawdzić GPU: brak biblioteki PyTorch.")

print("\n=== KONIEC RAPORTU ===")
```
